```python
student = {"Roll No": 101, "Name": "abc",    "Age": 20, "Department": "AI & ML","CGPA": 8.9}
print(student)
print("\nName:", student["Name"])
print("Department:", student.get("Department"))
student["City"] = "Coimbatore"
student["Mobile"] = "9876543210"
print("\nAfter Adding New Elements")
print(student)
student["Age"] = 21
student["CGPA"] = 9.1
print("\nAfter Updating Values")
print(student)
print("\nNumber of Elements:", len(student))
print("\nKeys", student.keys())
print("\nValues", student.values())
print("\nItems", student.items())
if "Name" in student:
    print("\nKey 'Name' Found")
else:
    print("\nKey Not Found")
removed = student.pop("Age")
print("\nRemoved Value:", removed)
del student["Department"]
print("\nAfter del", student)
student.clear()
print("\nAfter clear()",student)
```
