```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as pl
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([40,43,60,65,80])
model=LinearRegression()
model.fit(x,y)
pd=model.predict(x)
print("predicted mark",model.predict([[6]]))
pl.scatter(x,y,label="Actual data")
pl.plot(x,pd,label="predicted data")
pl.grid()
pl.xlabel("Study hours")
pl.ylabel("Marks scored")
pl.title("Student analysis")
pl.legend()
```
