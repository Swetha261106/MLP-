```python
import math

# Training data
X = [[1, 2], [2, 3], [3, 4], [4, 5]]
Y = [0, 0, 1, 1]

# Initial weights and bias
w1 = 0.5
w2 = 0.5
b = -2

lr = 0.1

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Training
for epoch in range(1000):
    for i in range(len(X)):
        x1 = X[i][0]
        x2 = X[i][1]

        z = x1 * w1 + x2 * w2 + b
        output = sigmoid(z)

        error = Y[i] - output
        gradient = error * output * (1 - output)

        w1 = w1 + lr * gradient * x1
        w2 = w2 + lr * gradient * x2
        b = b + lr * gradient




for i in range(len(X)):
    z = X[i][0] * w1 + X[i][1] * w2 + b
    output = sigmoid(z)

    print(X[i], "Actual =", Y[i], "Predicted =", round(output, 2))
```
