```python
from sklearn.linear_model import LinearRegression
import numpy as np
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([20,35,50,65,80])
model=LinearRegression()
model.fit(x,y)
predict_mark=model.predict([[6]])
print("predicted marks",predict_mark)
print("slope value",model.coef_)
print("intercept:",model.intercept_)
```

```python

x = [1, 2, 3, 4, 5]
y = [20, 40, 60, 80, 100]
n = len(x)
sum_x = sum(x)
sum_y = sum(y)
sum_xy = 0
sum_x2 = 0
for i in range(n):
 sum_xy += x[i] * y[i]
 sum_x2 += x[i] * x[i]
m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x * sum_x)
c = (sum_y - m * sum_x) / n
print("Slope (m):", m)
print("Intercept (c):", c)
new_x = 6
predicted_y = m * new_x + c
print("Predicted Marks for", new_x, "hours =", predicted_y)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as pl
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([40,43,60,65,80])
model=LinearRegression()
model.fit(x,y)
pd=model.predict(x)
print("predicted mark",model.predict([[6]]))
pl.scatter(x,y,label="Actual data")
pl.plot(x,pd,label="predicted data")
pl.grid()
pl.xlabel("Study hours")
pl.ylabel("Marks scored")
pl.title("Student analysis")
pl.legend()
```
